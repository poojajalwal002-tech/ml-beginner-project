## Step 1: Import Libraries

We'll use these libraries:
- **pandas**: For data manipulation
- **numpy**: For numerical operations
- **matplotlib** & **seaborn**: For visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for better-looking plots
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print("Libraries imported successfully!")
```

## Step 2: Load Sample Data

Let's create a simple dataset to explore. In real projects, you'd load data from a CSV file.

**This dataset represents:** Student performance data
- Study Hours: Hours spent studying
- Previous Score: Score from previous exam
- Attendance: Attendance percentage
- Final Score: The score we want to predict

```python
# Create sample dataset
np.random.seed(42)  # For reproducibility

data = {
    'Study_Hours': np.random.uniform(1, 8, 100),
    'Previous_Score': np.random.uniform(40, 95, 100),
    'Attendance': np.random.uniform(50, 100, 100),
    'Final_Score': np.random.uniform(30, 100, 100)
}

df = pd.DataFrame(data)
print("Dataset created!")
print(f"Shape: {df.shape} (rows, columns)")
```

## Step 3: First Look at the Data

Always start by looking at the first few rows to understand the structure.

```python
# View first few rows
print("First 5 rows of data:")
print(df.head())

print("\nLast 3 rows of data:")
print(df.tail(3))
```

## Step 4: Get Basic Information

Let's understand the data types and overall structure.

```python
# Data types and info
print("Data Types:")
print(df.dtypes)

print("\nDataframe Info:")
df.info()

print("\nShape of data:")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

## Step 5: Statistical Summary

Get descriptive statistics to understand the range and distribution of values.

```python
# Statistical summary
print("Statistical Summary:")
print(df.describe())

print("\nMore detailed statistics:")
print(df.describe(percentiles=[0.25, 0.5, 0.75]))
```

## Step 6: Check for Missing Values

Missing values can cause problems in ML. Let's find them!

```python
# Check missing values
print("Missing Values:")
missing = df.isnull().sum()
print(missing)

print("\nPercentage of missing values:")
print((df.isnull().sum() / len(df) * 100).round(2))

# Check if there are any missing values
if df.isnull().sum().sum() == 0:
    print("\n✅ Great! No missing values found.")
else:
    print("\n⚠️  There are missing values that need to be handled.")
```

## Step 7: Visualize the Data

A picture is worth a thousand words! Let's create visualizations.

```python
# Create subplots for each column
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for idx, column in enumerate(df.columns):
    row = idx // 2
    col = idx % 2
    axes[row, col].hist(df[column], bins=20, edgecolor='black', alpha=0.7)
    axes[row, col].set_title(f'Distribution of {column}')
    axes[row, col].set_xlabel(column)
    axes[row, col].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

print("Histograms show the distribution of each variable.")
```

## Step 8: Correlation Analysis

Which features are related to our target variable (Final_Score)?

```python
# Calculate correlation matrix
correlation = df.corr()
print("Correlation Matrix:")
print(correlation)

print("\nCorrelation with Final_Score:")
print(correlation['Final_Score'].sort_values(ascending=False))
```

## Step 9: Visualize Relationships

Create a correlation heatmap to see relationships visually.

```python
# Correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()

print("Heatmap shows how variables are related to each other.")
print("Values close to 1 mean strong positive relationship.")
print("Values close to -1 mean strong negative relationship.")
print("Values close to 0 mean no relationship.")
```

## Step 10: Scatter Plots

See how features relate to our target variable.

```python
# Scatter plots
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

features = ['Study_Hours', 'Previous_Score', 'Attendance']

for idx, feature in enumerate(features):
    axes[idx].scatter(df[feature], df['Final_Score'], alpha=0.6)
    axes[idx].set_xlabel(feature)
    axes[idx].set_ylabel('Final_Score')
    axes[idx].set_title(f'{feature} vs Final_Score')
    
    # Add a trend line
    z = np.polyfit(df[feature], df['Final_Score'], 1)
    p = np.poly1d(z)
    axes[idx].plot(df[feature].sort_values(), 
                   p(df[feature].sort_values()), 
                   "r--", alpha=0.8, linewidth=2)

plt.tight_layout()
plt.show()

print("The red dashed line shows the trend in the data.")
```

## Key Takeaways 🎯

By exploring data, we:
1. ✅ Understand the shape and size of our data
2. ✅ Identify data types (numeric, categorical, etc.)
3. ✅ Find missing or incorrect values
4. ✅ Discover relationships between variables
5. ✅ Get ideas for data preprocessing
6. ✅ Plan feature selection and engineering

## Questions to Always Ask 🤔

- How many rows and columns do we have?
- What are the data types?
- Are there missing values?
- What's the range of each variable?
- Which features relate to our target?
- Are there outliers or unusual patterns?
- Do we need to create new features?
