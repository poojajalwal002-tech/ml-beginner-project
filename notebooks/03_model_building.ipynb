# 03 - Model Building: Your First ML Models

Welcome to Notebook 03! Now that we have clean data, let's build our first machine learning models!

In this lesson:
1. Split data into training and testing sets
2. Build classification models
3. Build regression models
4. Compare model performance

## Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.preprocessing import StandardScaler

sns.set_style('whitegrid')
print("Libraries imported successfully!")
```

## Step 2: Load and Prepare Data for Classification

```python
# Create sample classification dataset
np.random.seed(42)

n_samples = 200
X = np.random.randn(n_samples, 2) * 2 + np.array([0, 0])
y = (X[:, 0] + X[:, 1] > 0).astype(int)  # Binary classification

# Add some noise
noise_idx = np.random.choice(n_samples, 20, replace=False)
y[noise_idx] = 1 - y[noise_idx]

df_class = pd.DataFrame(X, columns=['Feature1', 'Feature2'])
df_class['Target'] = y

print("Classification Dataset:")
print(df_class.head())
print(f"\nClass distribution:")
print(df_class['Target'].value_counts())

# Visualize
plt.figure(figsize=(8, 6))
scatter = plt.scatter(df_class['Feature1'], df_class['Feature2'], 
                      c=df_class['Target'], cmap='viridis', s=100, alpha=0.6)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Classification Dataset')
plt.colorbar(scatter)
plt.show()
```

## Step 3: Train-Test Split

**Why split?** To evaluate model performance on unseen data!

```python
# Prepare features and target
X_class = df_class[['Feature1', 'Feature2']]
y_class = df_class['Target']

# Split data: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X_class, y_class, test_size=0.2, random_state=42
)

print(f"Training set size: {len(X_train)}")
print(f"Testing set size: {len(X_test)}")
print(f"\nTraining set class distribution:")
print(y_train.value_counts())
print(f"\nTesting set class distribution:")
print(y_test.value_counts())
```

## Step 4: Train Classification Models

```python
# Model 1: Logistic Regression
print("=" * 50)
print("MODEL 1: Logistic Regression")
print("=" * 50)

lr_model = LogisticRegression(random_state=42)
lr_model.fit(X_train, y_train)
y_pred_lr = lr_model.predict(X_test)

print(f"Training accuracy: {lr_model.score(X_train, y_train):.4f}")
print(f"Testing accuracy: {lr_model.score(X_test, y_test):.4f}")
print(f"\nPredictions (first 10):")
print(y_pred_lr[:10])
```

```python
# Model 2: Decision Tree Classifier
print("\n" + "=" * 50)
print("MODEL 2: Decision Tree Classifier")
print("=" * 50)

dt_model = DecisionTreeClassifier(max_depth=5, random_state=42)
dt_model.fit(X_train, y_train)
y_pred_dt = dt_model.predict(X_test)

print(f"Training accuracy: {dt_model.score(X_train, y_train):.4f}")
print(f"Testing accuracy: {dt_model.score(X_test, y_test):.4f}")
print(f"\nFeature importances:")
for feature, importance in zip(['Feature1', 'Feature2'], dt_model.feature_importances_):
    print(f"  {feature}: {importance:.4f}")
```

```python
# Model 3: Random Forest Classifier
print("\n" + "=" * 50)
print("MODEL 3: Random Forest Classifier")
print("=" * 50)

rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)

print(f"Training accuracy: {rf_model.score(X_train, y_train):.4f}")
print(f"Testing accuracy: {rf_model.score(X_test, y_test):.4f}")
print(f"\nFeature importances:")
for feature, importance in zip(['Feature1', 'Feature2'], rf_model.feature_importances_):
    print(f"  {feature}: {importance:.4f}")
```

```python
# Model 4: K-Nearest Neighbors
print("\n" + "=" * 50)
print("MODEL 4: K-Nearest Neighbors (KNN)")
print("=" * 50)

knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
y_pred_knn = knn_model.predict(X_test)

print(f"Training accuracy: {knn_model.score(X_train, y_train):.4f}")
print(f"Testing accuracy: {knn_model.score(X_test, y_test):.4f}")
```

## Step 5: Evaluate and Compare Models

```python
# Create comparison dataframe
models_comparison = pd.DataFrame({
    'Model': ['Logistic Regression', 'Decision Tree', 'Random Forest', 'KNN'],
    'Training Accuracy': [
        lr_model.score(X_train, y_train),
        dt_model.score(X_train, y_train),
        rf_model.score(X_train, y_train),
        knn_model.score(X_train, y_train)
    ],
    'Testing Accuracy': [
        lr_model.score(X_test, y_test),
        dt_model.score(X_test, y_test),
        rf_model.score(X_test, y_test),
        knn_model.score(X_test, y_test)
    ]
})

print("\n" + "=" * 50)
print("MODEL COMPARISON")
print("=" * 50)
print(models_comparison)

# Visualize
fig, ax = plt.subplots(figsize=(10, 6))
x = np.arange(len(models_comparison))
width = 0.35

ax.bar(x - width/2, models_comparison['Training Accuracy'], width, label='Training', alpha=0.8)
ax.bar(x + width/2, models_comparison['Testing Accuracy'], width, label='Testing', alpha=0.8)

ax.set_xlabel('Model')
ax.set_ylabel('Accuracy')
ax.set_title('Model Comparison: Training vs Testing Accuracy')
ax.set_xticks(x)
ax.set_xticklabels(models_comparison['Model'], rotation=45, ha='right')
ax.legend()
ax.set_ylim([0, 1.1])

plt.tight_layout()
plt.show()
```

## Step 6: Detailed Evaluation Metrics

```python
# For the best model (Random Forest), calculate detailed metrics
print("\n" + "=" * 50)
print("DETAILED EVALUATION: Random Forest")
print("=" * 50)

print(f"\nAccuracy: {accuracy_score(y_test, y_pred_rf):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_rf):.4f}")
print(f"Recall: {recall_score(y_test, y_pred_rf):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred_rf):.4f}")

print("\n📊 Metric Explanations:")
print("- Accuracy: % of correct predictions")
print("- Precision: % of positive predictions that are correct")
print("- Recall: % of actual positives that are identified")
print("- F1-Score: Harmonic mean of precision and recall")
```

## Step 7: Regression Example

```python
# Create sample regression dataset
np.random.seed(42)
n_samples = 100
X_reg = np.random.randn(n_samples, 1) * 10
y_reg = 2.5 * X_reg.ravel() + np.random.randn(n_samples) * 5 + 10

df_reg = pd.DataFrame(X_reg, columns=['Feature'])
df_reg['Target'] = y_reg

print("\nRegression Dataset:")
print(df_reg.head())

# Visualize
plt.figure(figsize=(10, 6))
plt.scatter(df_reg['Feature'], df_reg['Target'], alpha=0.6, s=50)
plt.xlabel('Feature')
plt.ylabel('Target')
plt.title('Regression Dataset')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Split regression data
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    df_reg[['Feature']], df_reg['Target'], test_size=0.2, random_state=42
)

# Train regression models
print("\n" + "=" * 50)
print("REGRESSION MODELS")
print("=" * 50)

# Linear Regression
lr_reg = LinearRegression()
lr_reg.fit(X_train_reg, y_train_reg)
y_pred_lr_reg = lr_reg.predict(X_test_reg)

print("\nLinear Regression:")
print(f"R² Score: {r2_score(y_test_reg, y_pred_lr_reg):.4f}")
print(f"MAE: {mean_absolute_error(y_test_reg, y_pred_lr_reg):.4f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test_reg, y_pred_lr_reg)):.4f}")

# Decision Tree Regression
dt_reg = DecisionTreeRegressor(max_depth=5, random_state=42)
dt_reg.fit(X_train_reg, y_train_reg)
y_pred_dt_reg = dt_reg.predict(X_test_reg)

print("\nDecision Tree Regression:")
print(f"R² Score: {r2_score(y_test_reg, y_pred_dt_reg):.4f}")
print(f"MAE: {mean_absolute_error(y_test_reg, y_pred_dt_reg):.4f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test_reg, y_pred_dt_reg)):.4f}")

# Random Forest Regression
rf_reg = RandomForestRegressor(n_estimators=100, random_state=42)
rf_reg.fit(X_train_reg, y_train_reg)
y_pred_rf_reg = rf_reg.predict(X_test_reg)

print("\nRandom Forest Regression:")
print(f"R² Score: {r2_score(y_test_reg, y_pred_rf_reg):.4f}")
print(f"MAE: {mean_absolute_error(y_test_reg, y_pred_rf_reg):.4f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test_reg, y_pred_rf_reg)):.4f}")
```

```python
# Visualize regression predictions
plt.figure(figsize=(15, 5))

models_reg = [
    ('Linear Regression', lr_reg, y_pred_lr_reg),
    ('Decision Tree', dt_reg, y_pred_dt_reg),
    ('Random Forest', rf_reg, y_pred_rf_reg)
]

for idx, (name, model, y_pred) in enumerate(models_reg, 1):
    plt.subplot(1, 3, idx)
    
    # Plot actual vs predicted
    plt.scatter(y_test_reg, y_pred, alpha=0.6)
    
    # Plot perfect prediction line
    min_val = min(y_test_reg.min(), y_pred.min())
    max_val = max(y_test_reg.max(), y_pred.max())
    plt.plot([min_val, max_val], [min_val, max_val], 'r--', lw=2)
    
    plt.xlabel('Actual')
    plt.ylabel('Predicted')
    plt.title(name)
    r2 = r2_score(y_test_reg, y_pred)
    plt.text(0.05, 0.95, f'R² = {r2:.4f}', transform=plt.gca().transAxes,
            verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.show()
```

## Step 8: Make Predictions on New Data

```python
# Classification: Make predictions on new data
new_data_class = np.array([[1.5, 2.0], [-1.0, -0.5], [0.5, 0.5]])
predictions = rf_model.predict(new_data_class)
probabilities = rf_model.predict_proba(new_data_class)

print("\nNew Classification Predictions:")
print(f"New Data:\n{new_data_class}")
print(f"\nPredicted Classes: {predictions}")
print(f"\nPrediction Probabilities:")
for i, prob in enumerate(probabilities):
    print(f"  Sample {i}: Class 0: {prob[0]:.4f}, Class 1: {prob[1]:.4f}")
```

```python
# Regression: Make predictions on new data
new_data_reg = np.array([[-5], [0], [10], [20]])
predictions_reg = rf_reg.predict(new_data_reg)

print("\nNew Regression Predictions:")
print(f"New Features:\n{new_data_reg.ravel()}")
print(f"\nPredicted Values: {predictions_reg}")
```

## Key Takeaways 🎯

1. **Train-Test Split**: Always evaluate on unseen data
2. **Multiple Models**: Try different algorithms to find the best
3. **Metrics Matter**: 
   - Classification: Accuracy, Precision, Recall, F1-Score
   - Regression: R², MAE, RMSE
4. **Overfitting vs Underfitting**:
   - High train accuracy but low test = Overfitting
   - Low both = Underfitting
5. **Model Comparison**: Don't just pick the model with highest training accuracy!

## Common Models Quick Reference

| Model | Pros | Cons | Best For |
|-------|------|------|----------|
| Logistic Regression | Fast, interpretable | Linear only | Simple problems |
| Decision Tree | Interpretable | Overfitting | Simple problems |
| Random Forest | Powerful | Complex | Most problems |
| KNN | Simple | Slow, needs scaling | Small datasets |
